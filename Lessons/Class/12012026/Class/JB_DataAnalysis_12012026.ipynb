EX1:
Write a python function, named time_to_array_addition the function gets as input an array or a list. Check the type and adds 1 to each argument in the array list and prints the time it took.

```python
import numpy as np
import time

def time_to_array_addition(data):
    # 1.input type
    input_type = type(data)
    print(f"Input type: {input_type}")

    # 2. start time
    start_time = time.time()

    # 3. Perform the addition based on type
    if isinstance(data, np.ndarray):
        # NumPy Arr
        result = data + 1
    elif isinstance(data, list):
        # Standard list
        result = [x + 1 for x in data]
    else:
        print("Unsupported type.")
        return

    # 4. end-time
    end_time = time.time()
    # 5. Calc duration
    duration = end_time - start_time

    print(f"Time taken: {duration:.8f} sec")
    return result

# --- Test---
#list
my_list = list(range(1000000))
time_to_array_addition(my_list)

# NumPy array
my_array = np.arange(1000000)
time_to_array_addition(my_array)
```

1. Write a function that gets an argument n, checks if it's higher than 0 and than generates an array in that size of ones.

```python
def create_array(n):
  if n > 0 :
    return np.ones(n)
  return None

num = random.randint(-10,50)
arr = create_array(num)

print(f'n= {num}')
arr
```

2. Write a function that gets an argument n, checks if it's higher than 0 and than generates an array in that size between 0 to n with step 2.


```python
def arranged_array(n):
  if n > 0:
    return np.arange(0,n,2)
  return None

num = random.randint(-10,50)
arr = arranged_array(num)

#test
print(f'n= {num}')
arr
```

4. Write a python function that takes two arrays and checks if all the elements are equal

```python
def is_equal_arr(arr1, arr2):
  return np.array_equal(arr1,arr2)

#test
arr1 = [1,6,8,3,5]
arr2 = [6,8,3,5,1]

print(is_equal_arr(arr1,arr2))

arr3 = [1,1,3,6,9]
arr4 = [1,1,3,6,9]

print(is_equal_arr(arr1,arr3))
print(is_equal_arr(arr4,arr3))
```

5. Write a python function that takes two arrays and checks if they contain the same variables not neccesserly in the same order

```python
def is_same_orgs(arr1, arr2):
  return np.array_equal(np.sort(arr1), np.sort(arr2))


#test
arr1 = [1,6,8,3,5]
arr2 = [6,8,3,5,1]

print(is_same_orgs(arr1,arr2))

arr3 = [1,1,3,6,9]
arr4 = [1,1,3,6,9]

print(is_same_orgs(arr1,arr3))
print(is_same_orgs(arr4,arr3))
```
